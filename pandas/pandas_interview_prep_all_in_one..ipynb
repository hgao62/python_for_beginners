```python
import numpy as np
import time

# Create a large DataFrame
df_large = pd.DataFrame({
    'Trade Price': np.random.randint(90, 110, size=1_000_000),
    'theo': np.random.randint(95, 115, size=1_000_000)
})

# Using apply() (Slow)
start = time.time()
df_large['price_diff_apply'] = df_large.apply(lambda row: row['theo'] - row['Trade Price'], axis=1)
print("Apply() Time:", time.time() - start)

# Using vectorized operation (Fast)
start = time.time()
df_large['price_diff_vectorized'] = df_large['theo'] - df_large['Trade Price']
print("Vectorized Time:", time.time() - start)
```

1️⃣ What is a "View"?
A view means the sliced data still references the original DataFrame.
Any changes to the view will modify the original data.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({'A': np.arange(5), 'B': np.arange(5, 10)})
view_df = df[:3]  # Slicing without .copy() -> Possible View

view_df.loc[0, 'A'] = 100  # May modify df

# copy_df = df.loc[:3]
# df.loc[0, 'A'] = 100 
```

```python
view_df._is_view
```

2️⃣ What is a "Copy"?
A copy is an independent DataFrame that doesn't share memory with the original.




```python
df = pd.DataFrame({'A': np.arange(5), 'B': np.arange(5, 10)})

copy_df = df[:3].copy()  # ✅ Explicitly creating a copy

copy_df.loc[0, 'A'] = 100  # Won't affect df

print(df)  # ✅ Original df is unchanged!
```
